# Independent applicability flags

Build a training-photometry reference or load the frozen HSC reference. Flags do not depend on either trained network. The PCA projection is illustrative; the support decision uses the full standardized color space.

Run from the repository checkout after installing the package. Outputs go to `RUN_DIR`; original data are never overwritten.


```python
CONFIG_PATH = "configs/synthetic-optuna.yaml"
RUN_DIR = "runs/synthetic-optuna"
SHOW_PLOTS = True
```

```python
from pathlib import Path
from IPython.display import display
from turbopdz.workflow import run_stage, load_run

candidates = [Path.cwd(), *Path.cwd().parents]
PROJECT_ROOT = next((p for p in candidates if (p / "pyproject.toml").exists()), Path.cwd())
config_path = (PROJECT_ROOT / CONFIG_PATH).resolve()
run_dir = (PROJECT_ROOT / RUN_DIR).resolve()
print(f"config={config_path}\nrun={run_dir}")
```

```python
result = run_stage("applicability", config_path, run_dir)
display(result)
```

```python
from turbopdz.analysis import write_applicability_diagnostics
display(write_applicability_diagnostics(run_dir, show=SHOW_PLOTS))
```
