# PCA population and competing-solution checks

Examine a fitted PCA component and the galaxy populations it separates. Choose population bins and any physically motivated branch windows **before** interpreting the result. The synthetic defaults illustrate execution, not a spectral-desert interpretation.

Uses saved zML point-holdout predictions and preprocessing only. Reports counts, redshift-bin contributions to error/outliers, matched redshift-magnitude cells, and (when windows are supplied) branch confusion, Wilson count intervals, individual-PDZ branch probabilities and threshold sensitivity. PDFs/CSVs go to `RUN_DIR/plots/pca-population-PC*/` and `RUN_DIR/tables/pca-population-PC*/`.


```python
CONFIG_PATH = "configs/synthetic-optuna.yaml"
RUN_DIR = "runs/synthetic-optuna"
SHOW_PLOTS = True
COMPONENT = 1  # one-based; pick a fitted, varying component
REDSHIFT_EDGES = None  # None gives ten equal-width holdout bins
MAGNITUDE_EDGES = [19, 20, 21, 22, 23, 24, 25]
TAIL_FRACTION = 0.2
MIN_COMMON = 15
MIN_RATE_COUNT = 20
BRANCHES = None
INVESTIGATION_WINDOW = None
```

## Optional branch choices

Set `BRANCHES` to a mapping with `true_low`, `true_high`, `pdz_low`, `pdz_high` (two bounds each), `predicted_low_max`, and `predicted_high_min`. An illustrative choice is true-redshift windows `[0, 0.5]` / `[1.25, 1.75]`, PDZ windows `[0, 0.6]` / `[1.2, 1.8]`, and prediction cuts `0.6` / `1.2`. Adapt these bounds to physically motivated branches in the dataset; they are **not universal defaults**.

`INVESTIGATION_WINDOW` optionally highlights a chosen interval. A sparse spectroscopic histogram does not by itself establish photometric degeneracy or spectroscopic incompleteness. Mean bimodality is distinguished from probability in competing branches of individual PDZs. Two-peak shape checks are skipped on nonuniform grids; branch probabilities still honor the declared mass/density convention.

Overlap controls use only cells with at least `MIN_COMMON` objects in both groups. Empty overlap is reported without relaxing the binning. The outputs are exploratory conditional comparisons, not causal attribution or new calibration.


```python
from pathlib import Path
from IPython.display import display
from turbopdz.config import load_config, run_configs_compatible
from turbopdz.workflow import load_run
from turbopdz.analysis import write_pca_population_diagnostics

candidates = [Path.cwd(), *Path.cwd().parents]
PROJECT_ROOT = next((p for p in candidates if (p / "pyproject.toml").exists()), Path.cwd())
config_path = (PROJECT_ROOT / CONFIG_PATH).resolve()
run_dir = (PROJECT_ROOT / RUN_DIR).resolve()
_, _, saved = load_run(run_dir)
if not run_configs_compatible(saved, load_config(config_path), allow_path_changes=True):
    raise ValueError("Configuration differs from the saved scientific run")
print(f"config={config_path}\nrun={run_dir}")
```

```python
outputs = write_pca_population_diagnostics(
    run_dir, component=COMPONENT, redshift_edges=REDSHIFT_EDGES,
    magnitude_edges=MAGNITUDE_EDGES, tail_fraction=TAIL_FRACTION,
    min_common=MIN_COMMON, min_rate_count=MIN_RATE_COUNT,
    branches=BRANCHES, investigation_window=INVESTIGATION_WINDOW,
    show=SHOW_PLOTS,
)
display(outputs)
```
