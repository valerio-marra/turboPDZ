# Train and calibrate the error-scale proxy

Train the log-error network on development OOF residuals. Fit multiplicative calibration and the reliability reference CDF on calibration rows only. sigma_ml is an empirical error-scale proxy, not a calibrated posterior standard deviation.

Set `sigmaml.hpo.mode: search` for one development-only Optuna search minimizing held-fold log-error MAE, followed by a full-development refit at the selected epoch budget. `resume`, `fixed`, and `reuse_parameters` are also supported. No calibration or final-test objects enter the search.

Run from the repository checkout after installing the package. Outputs go to `RUN_DIR`; original data are never overwritten.


```python
CONFIG_PATH = "configs/synthetic-optuna.yaml"
RUN_DIR = "runs/synthetic-optuna"
SHOW_PLOTS = True
```

```python
from pathlib import Path
from IPython.display import display
from turbopdz.workflow import run_stage, load_run

candidates = [Path.cwd(), *Path.cwd().parents]
PROJECT_ROOT = next((p for p in candidates if (p / "pyproject.toml").exists()), Path.cwd())
config_path = (PROJECT_ROOT / CONFIG_PATH).resolve()
run_dir = (PROJECT_ROOT / RUN_DIR).resolve()
print(f"config={config_path}\nrun={run_dir}")
```

```python
result = run_stage("train-sigmaml", config_path, run_dir)
display(result)
```

```python
import json
from turbopdz.artifacts import load_manifest
display(load_manifest(run_dir / "sigmaml"))
display(json.loads((run_dir / "sigmaml/best_params.json").read_text()))
from turbopdz.analysis import write_training_diagnostics
display(write_training_diagnostics(run_dir, "sigmaml", show=SHOW_PLOTS))
```
