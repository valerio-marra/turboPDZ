# Train the optimized point estimate

Choose features, training loss, and Optuna selection objective in the configuration. Search, compatible resumption, and explicit parameter reuse are separate modes. Calibration and final-test rows never enter fitting or HPO.

Run from the repository checkout after installing the package. Outputs go to `RUN_DIR`; original data are never overwritten.


```python
CONFIG_PATH = "configs/synthetic-optuna.yaml"
RUN_DIR = "runs/synthetic-optuna"
SHOW_PLOTS = True
```

```python
from pathlib import Path
from IPython.display import display
from turbopdz.workflow import run_stage, load_run

candidates = [Path.cwd(), *Path.cwd().parents]
PROJECT_ROOT = next((p for p in candidates if (p / "pyproject.toml").exists()), Path.cwd())
config_path = (PROJECT_ROOT / CONFIG_PATH).resolve()
run_dir = (PROJECT_ROOT / RUN_DIR).resolve()
print(f"config={config_path}\nrun={run_dir}")
```

```python
result = run_stage("train-zml", config_path, run_dir)
display(result)
```

```python
import json
display(json.loads((run_dir / "zml" / "best_params.json").read_text()))
from turbopdz.analysis import write_training_diagnostics
display(write_training_diagnostics(run_dir, "zml", show=SHOW_PLOTS))
```
