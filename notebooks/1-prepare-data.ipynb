# Prepare and validate data

Adapt a dataset, preserve its native PDZ grid and probability convention, and save one persistent object-level split plan. Edit the experiment configuration and use an adapter for dataset-specific ingestion; keep that logic separate from the training core.

Run from the repository checkout after installing the package. Outputs go to `RUN_DIR`; original data are never overwritten.


```python
CONFIG_PATH = "configs/synthetic-optuna.yaml"
RUN_DIR = "runs/synthetic-optuna"
SHOW_PLOTS = True
```

```python
from pathlib import Path
from IPython.display import display
from turbopdz.workflow import run_stage, load_run

candidates = [Path.cwd(), *Path.cwd().parents]
PROJECT_ROOT = next((p for p in candidates if (p / "pyproject.toml").exists()), Path.cwd())
config_path = (PROJECT_ROOT / CONFIG_PATH).resolve()
run_dir = (PROJECT_ROOT / RUN_DIR).resolve()
print(f"config={config_path}\nrun={run_dir}")
```

```python
result = run_stage("prepare", config_path, run_dir)
display(result)
```

```python
dataset, splits, config = load_run(run_dir)
display(dataset.observables.head())
print({"development": len(splits.development), "calibration": len(splits.calibration), "final_test": len(splits.test), "oof_folds": splits.n_folds})
from turbopdz.analysis import write_data_diagnostics
display(write_data_diagnostics(run_dir, show=SHOW_PLOTS))
```
